#### Ejemplos de if, if else, if elif

#### Ejemplo de if

```python
# Ejemplo if
edad = 20  # Asigna el valor entero 20 a la variable edad
if edad >= 18:  # Evalúa si la edad es mayor o igual a 18 usando el operador relacional >=
    print("Eres mayor de edad.")  # Imprime el mensaje si la condición anterior es verdadera
```

```python
# Ejemplo If else
numero = 15  # Se inicializa la variable numero con el valor 15
if numero % 2 == 0:  # Usa el operador aritmético % para obtener el resto y == para comparar si es cero
    print("El número es par.")  # Se ejecuta si el resto de la división entre 2 es exactamente 0
else:  # Define el bloque alternativo si la condición del if resulta falsa
    print("El número es impar.")  # Informa que el número no es divisible exactamente entre 2
```

```python
# Ejemplo If else
numero = 0  # Se inicializa la variable numero con el valor 15
if numero % 2 == 0:  # Usa el operador aritmético % para obtener el resto y == para comparar si es cero
    print("El número es par.")  # Se ejecuta si el resto de la división entre 2 es exactamente 0
else:  # Define el bloque alternativo si la condición del if resulta falsa
    print("El número es impar.")  # Informa que el número no es divisible exactamente entre 2
```

```python
# Ejemplo if elif

valor = -5  # Asigna el número entero -5 a la variable valor
if valor > 0:  # Verifica si el valor es mayor que cero (positivo)
    print("Es positivo.")  # Muestra el mensaje si el número es superior a 0
elif valor < 0:  # Evalúa si el valor es menor que cero (negativo) solo si el if falló
    print("Es negativo.")  # Muestra el mensaje si el número es inferior a 0
else:  # Se ejecuta únicamente si no se cumplió ninguna de las condiciones anteriores
    print("Es cero.")  # Informa que el valor es exactamente 0
```

```python
# ejemplo if elif
# Ejemplo: Categorizar un número usando múltiples comparaciones
numero = int(input("Ingrese un número: "))
if numero > 0:
    print("El número es positivo.")
elif numero < 0:
    print("El número es negativo.")
else:
    print("El número es cero.")
```
